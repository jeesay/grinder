```python
merged = pd.merge(prcss,edges)
merged
```

```python
# cleaned
imports = merged[merged['rlnPipeLineProcessTypeLabel'] == 'relion.import.movies'].copy()
mc = merged[merged['rlnPipeLineProcessName'].str.contains('Motion')]
filters = (merged['rlnPipeLineProcessName'].str.contains('CtfFind')) & (merged['rlnPipeLineEdgeProcess'].str.contains('Pick'))
ctfs = merged[filters].copy()
picks = merged[(merged['rlnPipeLineProcessName'].str.contains('Pick')) & (merged['rlnPipeLineEdgeProcess'].str.contains('Extract'))]
select = merged[merged['rlnPipeLineProcessName'].str.contains('Select')]
extract =  merged[(merged['rlnPipeLineProcessName'].str.contains('Extract')) & (merged['rlnPipeLineEdgeProcess'].str.contains('Class2D'))]
cluster = merged[(merged['rlnPipeLineProcessName'].str.contains('Class')) & (merged['rlnPipeLineEdgeProcess'].str.contains('Select'))]
initio = merged[(merged['rlnPipeLineProcessName'].str.contains('Initial')) & (merged['rlnPipeLineEdgeProcess'].str.contains('Class3D'))]
cleaned = pd.concat([imports,mc,ctfs,picks,select,extract,cluster,initio]) # without 
```

```python
cluster
```

```python
def set_name(column):
    def foo(x):
        tools = x['rlnPipeLineProcessTypeLabel'].split('.')
        tool = ''
        if len(tools) == 4:
            tool = f'{tools[-2]}{tools[-1].capitalize()}_'
        elif len(tools) == 3:
            tool = tools[-1] + '_'
        _name = x[column].replace('/job',f'_{tool}')
        return _name[:-1]
    return foo
    
cleaned['ProcessName'] = cleaned.apply(set_name('rlnPipeLineProcessName'),axis=1)
cleaned
```

```python
import pandas as pd
from collections import defaultdict

def extract_branches(df):
    # Create a dictionary to hold the graph structure
    graph = defaultdict(list)
    for _, row in df.iterrows():
        graph[row['rlnPipeLineProcessName']].append(row['rlnPipeLineEdgeProcess'])

    # Function to perform a depth-first search and extract branches
    def dfs(node, path, branches):
        path.append(node)
        if node not in graph or not graph[node]:
            branches.append(path.copy())
        else:
            for next_node in graph[node]:
                dfs(next_node, path, branches)
        path.pop()

    # Extract branches
    branches = []
    # Assuming there is at least one starting node with no incoming edges
    # You might need to adjust this part based on your specific data
    starting_nodes = [node for node in df['rlnPipeLineProcessName'] if node not in df['rlnPipeLineEdgeProcess'].values]
    for node in set(starting_nodes):
        print('node',node)
        dfs(node, [], branches)

    return branches
```

```python
branches = extract_branches(cleaned)
```

```python
for b in branches:
    print(b)
```

```python
sorted_branches = sorted(branches,key=lambda x: len(x))
```

```python
for b in sorted_branches:
    print(b)
```

```python
# main
count = len(sorted_branches) - 1
for j,op in enumerate(sorted_branches[-1]):
    print(f'commit id:"{op.replace('/job','_')[:-1]}"')
    for i in range(j,len(sorted_branches) - 1):
        # print(j,i)
        tool = sorted_branches[i][j]
        if tool != op:
            print(f'branch b{count:03d}')
            for k in range(j,len(sorted_branches[i])):
                tool = sorted_branches[i][k]
                print(f'commit id:"{tool.replace('/job','_')[:-1]}"')
            print(f'checkout main')
            count -= 1
```
